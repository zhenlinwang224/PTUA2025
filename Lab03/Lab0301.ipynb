```python
def count(word, letter):
    count = 0  
    for char in word: 
        if char == letter:  
            count += 1  
    return count  


print(count("banana", "a"))  
print(count("apple", "p"))   
```

```python
def is_palindrome(word):
    return word == word[::-1]
print(is_palindrome("noon")) 
print(is_palindrome("redivider")) 
print(is_palindrome("banana"))  
print(is_palindrome("a"))  
print(is_palindrome(""))  
```

```python
def any_lowercase1(s):
    for c in s:
        if c.islower():
            return True
        else:
            return False
print(any_lowercase1("Apple"))  
print(any_lowercase1("apple"))  
print(any_lowercase1("APPLEa"))
```

```python
def any_lowercase2(s):
    for c in s:
        if 'c'.islower():
            return 'True'
        else:
            return 'False'
print(any_lowercase2("APPLE"))  
print(any_lowercase2("apple")) 
```

```python
def any_lowercase3(s):
    for c in s:
        flag = c.islower()  
    return flag
print(any_lowercase3("APPLEa")) 
print(any_lowercase3("Apple"))  
print(any_lowercase3("APPLES")) 
```

```python
def any_lowercase4(s):
    flag = False
    for c in s:
        flag = flag or c.islower()  
    return flag
print(any_lowercase4("APPLE"))  
print(any_lowercase4("Apple"))  
print(any_lowercase4("APPLEa")) 
```

```python
def any_lowercase5(s):
    for c in s:
        if not c.islower():  
            return False
    return True 
print(any_lowercase5("apple"))  
print(any_lowercase5("Apple")) 
print(any_lowercase5("appleA")) 
```

```python
def rotate_word(word, shift):
    result = ""  

    for char in word:
        if char.isalpha():  
            if char.islower(): 
                new_char = chr((ord(char) - ord('a') + shift) % 26 + ord('a'))
            else:  
                new_char = chr((ord(char) - ord('A') + shift) % 26 + ord('A'))
        else:
            new_char = char  

        result += new_char  

    return result


print(rotate_word("cheer", 7))  
```

```python
print(rotate_word("Hello, World!", 5))
```
