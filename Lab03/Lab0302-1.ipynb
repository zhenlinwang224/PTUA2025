```python
import random
import math


def estimate_pi(convergence_criterion, max_draws):
    inside_circle = 0
    total_draws = 0
    true_pi = math.pi  # True value of Pi

    while total_draws < max_draws:
        x, y = random.random(), random.random()  
        if x**2 + y**2 <= 1: 
            inside_circle += 1

        total_draws += 1
        pi_estimate = (inside_circle / total_draws) * 4 
        percentage_difference = abs((pi_estimate - true_pi) / true_pi)

        
        if percentage_difference < convergence_criterion:
            print(f"Converged at {total_draws} draws. Estimated Pi: {pi_estimate:.6f}")
            return pi_estimate, total_draws, False  

    
    print(f"Draw limit exceeded ({max_draws}). Estimated Pi: {pi_estimate:.6f}")
    return pi_estimate, total_draws, True  


convergence_criterion = float(input("Enter the convergence criterion (e.g., 0.01, 0.001): "))
max_draws = int(input("Enter the sentinel value (max number of draws): "))


pi_value, draws, sentinel_reached = estimate_pi(convergence_criterion, max_draws)
```
